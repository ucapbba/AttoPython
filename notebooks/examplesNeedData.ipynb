## 1. MNIST digit classification

From `MLexamples/Classification/MNIST/RunSGDClassifier.py`. Trains an `SGDClassifier` on MNIST digit images.

**Requires data:** the cell below (from `SaveMNISTdata.py`) downloads `mnist_784` from OpenML and caches it as `mnist_784.pkl` the first time it's run — this needs internet access and can take a while. If the pickle already exists it's reused.

```python
import os
from sklearn.datasets import fetch_openml
from MLexamples.Classification.MNIST.DataFunctions import SaveBunchPickle

mnistPickleDir = "/MLexamples/Classification/MNIST/"
mnistPicklePath = os.getcwd() + mnistPickleDir + "mnist_784.pkl"

if os.path.exists(mnistPicklePath):
    print("Found existing", mnistPicklePath, "- skipping download")
else:
    print("Downloading mnist_784 from OpenML (first run only)...")
    mnist = fetch_openml("mnist_784", version=1, as_frame=True)
    SaveBunchPickle("mnist_784.pkl", mnistPickleDir, mnist)
    print("Saved", mnistPicklePath)
```

```python
from MLexamples.Classification.MNIST.DataFunctions import LoadBunchPickle, PrintDataDetails
from MLexamples.Classification.MNIST.MLFunctions import MLFunctions

mnistData = LoadBunchPickle("mnist_784.pkl", "/MLexamples/Classification/MNIST/")
data = mnistData["data"]
numbers = mnistData["target"]

PrintDataDetails(data)
PrintDataDetails(numbers)

truncate = 10000
images = data.values[:truncate]
numbers = numbers[:truncate]

ML = MLFunctions(images, numbers, 0.01)
ML.MakePredictions(images, numbers)

for prediction, y in zip(ML.predictions, ML.y_test):
    print("prediction = " + str(prediction) + " ### actual = " + str(y))

print("Getting performance measures")
ML.GetPerformanceMeasures()
```

## 2. ARC asymmetry calculation

From `Source/MachineLearning/ARC/Calc_Assym.py`. Calculates asymmetry values across a batch of ARC job output files using `AsymCalculator`.

**Requires data:** job output files under `Data/job/` — not included in this repository, so this cell won't find any input and will just print that no contour helpers were populated.

```python
from Source.MachineLearning.ARC.AsymCalculator import AsymCalculator
from Source.MachineLearning.ARC.AsymCalculatorIO import AsymCalculatorIO

xRange, yRange = 500, 500
_min, _max = 1e-6, 1e2
path = "/Data/job/"
jobNum = 41044
minTask, maxTask = 1, 50

calc = AsymCalculator(path, jobNum, minTask, maxTask, xRange, yRange, _min, _max)
calc.CreateContourHelpers()
calc.CalculateAsymmetries()

print("Calculation finished, output calculations")
io = AsymCalculatorIO(calc)
del calc
print("writing to file......")
io.printToScreen()
io.printToFile()
print("Done")
```
